```python
from pathlib import Path
import re

root = Path(
    "/content/models/research/"
    "Tensorflow-Object-Detection-API-With-Custom-Dataset"
)

target = root / "pipeline.config"
source = (
    root / "pretrained_model/"
    "ssd_mobilenet_v2_320x320_coco17_tpu-8/"
    "pipeline.config"
)

# 先備份目前設定
backup = root / "pipeline.config.before_anchor_fix.bak"
backup.write_text(target.read_text())

def extract_block(content, block_name):
    match = re.search(r"\b" + re.escape(block_name) + r"\s*\{", content)
    if not match:
        raise ValueError(f"找不到區塊：{block_name}")

    open_brace = content.find("{", match.start())
    depth = 0

    for i in range(open_brace, len(content)):
        if content[i] == "{":
            depth += 1
        elif content[i] == "}":
            depth -= 1
            if depth == 0:
                return content[match.start():i + 1], match.start(), i + 1

    raise ValueError(f"區塊括號不完整：{block_name}")

original = target.read_text()
pretrained = source.read_text()

old_block, start, end = extract_block(original, "anchor_generator")
new_block, _, _ = extract_block(pretrained, "anchor_generator")

if "ssd_anchor_generator" not in new_block:
    raise ValueError("來源設定沒有 ssd_anchor_generator，停止修改。")

updated = original[:start] + new_block + original[end:]
target.write_text(updated)

print("已備份：", backup)
print("已替換 anchor_generator 區塊。")
print("目前模型類型：",
      re.search(r'type:\s*"([^"]+)"', updated).group(1))
print("保留 num_classes: 1：", "num_classes: 1" in updated)
print("使用 ssd_anchor_generator：",
      "ssd_anchor_generator" in extract_block(updated, "anchor_generator")[0])
print("仍含 multiscale_anchor_generator：",
      "multiscale_anchor_generator" in
      extract_block(updated, "anchor_generator")[0])
```

```python
from pathlib import Path
import re
import shutil

root = Path(
    "/content/models/research/"
    "Tensorflow-Object-Detection-API-With-Custom-Dataset"
)

current = root / "pipeline.config"
pretrained = (
    root / "pretrained_model/"
    "ssd_mobilenet_v2_320x320_coco17_tpu-8/"
    "pipeline.config"
)

backup = root / "pipeline.config.before_predictor_fix.bak"

# 先備份目前設定
shutil.copy2(current, backup)

current_text = current.read_text()
pretrained_text = pretrained.read_text()

def extract_block(text, name):
    match = re.search(r"\b" + re.escape(name) + r"\s*\{", text)
    if not match:
        raise ValueError(f"找不到 {name} 區塊")

    start = match.start()
    brace = text.find("{", match.start())
    depth = 0

    for i in range(brace, len(text)):
        if text[i] == "{":
            depth += 1
        elif text[i] == "}":
            depth -= 1
            if depth == 0:
                return start, i + 1, text[start:i + 1]

    raise ValueError(f"{name} 區塊括號不完整")

c_start, c_end, _ = extract_block(current_text, "box_predictor")
_, _, pretrained_block = extract_block(pretrained_text, "box_predictor")

updated = (
    current_text[:c_start]
    + pretrained_block
    + current_text[c_end:]
)

current.write_text(updated)

print("已備份：", backup)
print("已替換 box_predictor 區塊。")
print("目前 num_classes：", re.search(
    r"num_classes:\s*\d+", updated
).group(0))
print("使用 convolutional_box_predictor：",
      "convolutional_box_predictor" in updated)
print("仍使用 weight_shared_convolutional_box_predictor：",
      "weight_shared_convolutional_box_predictor" in updated)
```

```python
import os
import glob
import pandas as pd
import xml.etree.ElementTree as ET


def xml_to_csv(path):
    xml_list = []
    for xml_file in glob.glob(path + '/*.xml'):
        tree = ET.parse(xml_file)
        root = tree.getroot()
        for member in root.findall('object'):
            value = (root.find('filename').text,
                     int(root.find('size')[0].text),
                     int(root.find('size')[1].text),
                     member[0].text,
                     int(member[4][0].text),
                     int(member[4][1].text),
                     int(member[4][2].text),
                     int(member[4][3].text)
                     )
            xml_list.append(value)
    column_name = ['filename', 'width', 'height', 'class', 'xmin', 'ymin', 'xmax', 'ymax']
    xml_df = pd.DataFrame(xml_list, columns=column_name)
    return xml_df


def main():
    image_path = os.path.join(os.getcwd(), 'annotations')
    xml_df = xml_to_csv(image_path)
    xml_df.to_csv('raccoon_labels.csv', index=None)
    print('Successfully converted xml to csv.')


main()
```

```python
"""
Usage:
  # From tensorflow/models/
  # Create train data:
  python generate_tfrecord.py --csv_input=data/train_labels.csv  --output_path=train.record

  # Create test data:
  python generate_tfrecord.py --csv_input=data/test_labels.csv  --output_path=test.record
"""
from __future__ import division
from __future__ import print_function
from __future__ import absolute_import

import os
import io
import sys
import pandas as pd
import tensorflow as tf

from PIL import Image
from object_detection.utils import dataset_util
from collections import namedtuple, OrderedDict

# 停用 TF2 行為以相容舊版代碼
tf.compat.v1.disable_eager_execution()

flags = tf.compat.v1.flags

# 安全地定義參數，避免重複執行單元格時報錯
def define_flag_safely(define_func, name, default, help_text):
    if name not in flags.FLAGS:
        define_func(name, default, help_text)

define_flag_safely(flags.DEFINE_string, 'csv_input', '', 'Path to the CSV input')
define_flag_safely(flags.DEFINE_string, 'output_path', '', 'Path to output TFRecord')
define_flag_safely(flags.DEFINE_string, 'image_dir', '', 'Path to images')

FLAGS = flags.FLAGS

# TO-DO replace this with label map
def class_text_to_int(row_label):
    if row_label == 'raccoon' or row_label == 'apple':
        return 1
    else:
        return None


def split(df, group):
    data = namedtuple('data', ['filename', 'object'])
    gb = df.groupby(group)
    return [data(filename, gb.get_group(x)) for filename, x in zip(gb.groups.keys(), gb.groups)]


def create_tf_example(group, path):
    with tf.compat.v1.gfile.GFile(os.path.join(path, '{}'.format(group.filename)), 'rb') as fid:
        encoded_jpg = fid.read()
    encoded_jpg_io = io.BytesIO(encoded_jpg)
    image = Image.open(encoded_jpg_io)
    width, height = image.size

    filename = group.filename.encode('utf8')
    image_format = b'jpg'
    xmins = []
    xmaxs = []
    ymins = []
    ymaxs = []
    classes_text = []
    classes = []

    for index, row in group.object.iterrows():
        xmins.append(row['xmin'] / width)
        xmaxs.append(row['xmax'] / width)
        ymins.append(row['ymin'] / height)
        ymaxs.append(row['ymax'] / height)
        classes_text.append(row['class'].encode('utf8'))
        classes.append(class_text_to_int(row['class']))

    tf_example = tf.train.Example(features=tf.train.Features(feature={
        'image/height': dataset_util.int64_feature(height),
        'image/width': dataset_util.int64_feature(width),
        'image/filename': dataset_util.bytes_feature(filename),
        'image/source_id': dataset_util.bytes_feature(filename),
        'image/encoded': dataset_util.bytes_feature(encoded_jpg),
        'image/format': dataset_util.bytes_feature(image_format),
        'image/object/bbox/xmin': dataset_util.float_list_feature(xmins),
        'image/object/bbox/xmax': dataset_util.float_list_feature(xmaxs),
        'image/object/bbox/ymin': dataset_util.float_list_feature(ymins),
        'image/object/bbox/ymax': dataset_util.float_list_feature(ymaxs),
        'image/object/class/text': dataset_util.bytes_list_feature(classes_text),
        'image/object/class/label': dataset_util.int64_list_feature(classes),
    }))
    return tf_example


def main(_):
    # 檢查是否在 Colab 中直接執行，若無參數則設定預設測試路徑
    csv_input = FLAGS.csv_input if FLAGS.csv_input else 'raccoon_labels.csv'
    output_path_val = FLAGS.output_path if FLAGS.output_path else 'train.record'
    image_dir_val = FLAGS.image_dir if FLAGS.image_dir else 'images/train'

    writer = tf.compat.v1.python_io.TFRecordWriter(output_path_val)
    path = os.path.join(image_dir_val)
    examples = pd.read_csv(csv_input)
    grouped = split(examples, 'filename')
    for group in grouped:
        try:
            tf_example = create_tf_example(group, path)
            writer.write(tf_example.SerializeToString())
        except Exception as e:
            print('Error processing image {}: {}'.format(group.filename, e))

    writer.close()
    output_path_full = os.path.join(os.getcwd(), output_path_val)
    print('Successfully created the TFRecords: {}'.format(output_path_full))


if __name__ == '__main__':
    # 避免系統參數干擾，手動初始化並解析參數
    if not FLAGS.is_parsed():
        FLAGS(sys.argv[:1])
    tf.compat.v1.app.run()
```

```python

from pathlib import Path
import tf_slim

file_path = (
    Path(tf_slim.__file__).parent
    / "data"
    / "tfexample_decoder.py"
)

source = file_path.read_text()
backup_path = file_path.with_suffix(".py.bak")

# 保留原始檔案，避免修改後無法還原
if not backup_path.exists():
    backup_path.write_text(source)
    print("已備份原始檔案：", backup_path)
else:
    print("原始備份已存在：", backup_path)

# 只在檔案尚未匯入 tensorflow as tf 時加入
if "import tensorflow as tf" not in source:
    marker = "from __future__ import print_function"
    if marker not in source:
        raise RuntimeError("找不到預期的插入位置，未修改檔案。")

    source = source.replace(
        marker,
        marker + "\n\nimport tensorflow as tf",
        1
    )
    file_path.write_text(source)
    print("已加入 import tensorflow as tf")
else:
    print("檔案已經有 tensorflow as tf，未修改。")
```

```python
from pathlib import Path
import re
from object_detection.utils import config_util

config_path = Path(
    "/content/models/research/"
    "Tensorflow-Object-Detection-API-With-Custom-Dataset/"
    "pipeline.config"
)

# 先備份目前設定
backup_path = config_path.with_name("pipeline.config.before_v2.bak")
if not backup_path.exists():
    backup_path.write_text(config_path.read_text())

text = config_path.read_text()

# 找出並修改設定；若原本沒有這一行，就補進 train_config 區塊
pattern = r"(?m)^(\s*fine_tune_checkpoint_version\s*:\s*)\S+"
if re.search(pattern, text):
    text = re.sub(pattern, r"\g<1>V2", text, count=1)
else:
    match = re.search(r"(?m)^(\s*train_config\s*\{)", text)
    if not match:
        raise ValueError("找不到 train_config 區塊，先不要修改檔案。")

    insert_at = match.end()
    text = (
        text[:insert_at]
        + "\n  fine_tune_checkpoint_version: V2"
        + text[insert_at:]
    )

config_path.write_text(text)

# 重新讀取，確認實際設定
configs = config_util.get_configs_from_pipeline_file(str(config_path))
print("目前版本設定：", configs["train_config"].fine_tune_checkpoint_version)
```

```python
from pathlib import Path
import xml.etree.ElementTree as ET
import pandas as pd

base = Path(
    "/content/models/research/"
    "Tensorflow-Object-Detection-API-With-Custom-Dataset"
)

for split_name in ["train", "test"]:
    folder = base / "images" / split_name
    rows = []

    for xml_path in sorted(folder.glob("*.xml")):
        root = ET.parse(xml_path).getroot()

        filename = root.findtext("filename")
        size = root.find("size")
        width = int(size.findtext("width"))
        height = int(size.findtext("height"))

        for obj in root.findall("object"):
            box = obj.find("bndbox")

            rows.append({
                "filename": filename,
                "width": width,
                "height": height,
                "class": obj.findtext("name").strip(),
                "xmin": int(float(box.findtext("xmin"))),
                "ymin": int(float(box.findtext("ymin"))),
                "xmax": int(float(box.findtext("xmax"))),
                "ymax": int(float(box.findtext("ymax"))),
            })

    output = base / "data" / f"{split_name}_labels.csv"
    pd.DataFrame(
        rows,
        columns=[
            "filename", "width", "height", "class",
            "xmin", "ymin", "xmax", "ymax"
        ]
    ).to_csv(output, index=False)

    print(f"{split_name}：{len(rows)} 筆標註 → {output}")
```

```python
from pathlib import Path
import shutil

file_path = Path(
    "/content/models/research/object_detection/model_lib_v2.py"
)

# 先備份，避免修改後無法還原
backup_path = file_path.with_name("model_lib_v2.py.before_keras3.bak")
if not backup_path.exists():
    shutil.copy2(file_path, backup_path)

lines = file_path.read_text().splitlines(keepends=True)

# 找出不相容的呼叫
matches = [
    i for i, line in enumerate(lines)
    if "tf.keras.backend.set_learning_phase(False)" in line
]

if not matches:
    print("找不到目標程式碼，先不要繼續修改。")
else:
    # 只修改這次錯誤涉及的呼叫
    i = matches[0]
    indent = lines[i][:len(lines[i]) - len(lines[i].lstrip())]

    lines[i:i+1] = [
        indent + "if hasattr(tf.keras.backend, 'set_learning_phase'):\n",
        indent + "    tf.keras.backend.set_learning_phase(False)\n"
    ]

    file_path.write_text("".join(lines))
    print("已完成相容性修正。")
    print("原始檔案備份位置：", backup_path)
    print("其他相同呼叫的數量：", len(matches) - 1)
```
